Welcome to New York City, one of the most-visited cities in the world. There are many Airbnb listings in New York City to meet the high demand for temporary lodging for travelers, which can be anywhere between a few nights to many months. In this project, we will take a closer look at the New York Airbnb market by combining data from multiple file types like `.csv`, `.tsv`, and `.xlsx`.

Recall that **CSV**, **TSV**, and **Excel** files are three common formats for storing data. 
Three files containing data on 2019 Airbnb listings are available to you:

**data/airbnb_price.csv**
This is a CSV file containing data on Airbnb listing prices and locations.
- **`listing_id`**: unique identifier of listing
- **`price`**: nightly listing price in USD
- **`nbhood_full`**: name of borough and neighborhood where listing is located

**data/airbnb_room_type.xlsx**
This is an Excel file containing data on Airbnb listing descriptions and room types.
- **`listing_id`**: unique identifier of listing
- **`description`**: listing description
- **`room_type`**: Airbnb has three types of rooms: shared rooms, private rooms, and entire homes/apartments

**data/airbnb_last_review.tsv**
This is a TSV file containing data on Airbnb host names and review dates.
- **`listing_id`**: unique identifier of listing
- **`host_name`**: name of listing host
- **`last_review`**: date when the listing was last reviewed

### Library Set-Up

```python
# Import necessary packages
import pandas as pd
import numpy as np
```

### Ingest Data

```python
#File information
fileNames = ['airbnb_last_review.tsv','airbnb_price.csv','airbnb_room_type.xlsx']
```

```python
pricesDF = pd.read_csv( fileNames[1] )
roomTypeDF = pd.read_excel( fileNames[2])
```

```python
# File Exploration
# read fileName[0] : airbnb_last_review.tsv

with open( fileNames[0], 'r') as mFile:
    print(mFile.readline())
    print(mFile.readline())
    print(mFile.readline())
```

```python
lrData = np.loadtxt( fileNames[0], dtype = 'str', delimiter = '\t', skiprows = 0)
```

```python
lrData[:4]
```

```python
#data[0] = column names
#data[1:] = row values
fColNames = lrData[0]

lastReviewDF = pd.DataFrame( data = lrData[1:], columns = fColNames )
```

```python
lastReviewDF['listing_id'] = lastReviewDF['listing_id'].astype(int)
```

```python
lastReviewDF.head()
```

### Tasks

**Task A:** What are the dates of the earliest and most recent reviews? Store these values as two separate variables with your preferred names.

```python
lastReviewDF['last_review'].describe()
```

```python
# last_review stored as string
# update to datetime
lastReviewDF['lr_date'] = pd.to_datetime(lastReviewDF['last_review'], format = '%B %d %Y')
```

```python
lastReviewDF['lr_date'][:7]
```

```python
lastReviewDF = lastReviewDF.sort_values( by = 'lr_date', ascending = False)
lastReviewDF.head()
```

```python
#Sorted newest to oldest. Positions are 0 and -1
oldestReviewDate = lastReviewDF['lr_date'].iloc[-1]
newestReviewDate = lastReviewDF['lr_date'].iloc[0]
```

```python
print("Oldest review date ", oldestReviewDate )
print("Newest review date ", newestReviewDate )
```

**Task B:** How many of the listings are private rooms? Save this into any variable.

```python
roomTypeDF.head()
```

```python
roomTypeDF.shape
```

```python
roomTypeDF['room_type'].unique()
```

```python
# Multiple uppercase/lowercase combinations
# Standardizig room_type to all lower case

roomTypeDF['room_type'] = roomTypeDF['room_type'].str.lower()
```

```python
prString = 'private room'

privateRoomListingCount = roomTypeDF[ roomTypeDF['room_type'] == prString ]['room_type'].count()
```

```python
print(privateRoomListingCount)
```

**Task C:** What is the average listing price? Round to the nearest two decimal places and save into a variable.

```python
pricesDF.head()
```

```python
# Drop "dollars" from columns
pricesDF['price_dol'] = pricesDF['price'].str.replace(" dollars","")

#Transform column into integer
pricesDF['price_dol'] = pricesDF['price_dol'].astype(int)
```

```python
pricesDF.head()
```

```python
pricesDF['price_dol'].describe()
```

```python
avgListingPrice = round( pricesDF['price_dol'].mean(), 2)
avgListingPrice
```

**Task D:** Combine the new variables into one DataFrame called `review_dates` with four columns in the following order: `first_reviewed`, `last_reviewed`, `nb_private_rooms`, and `avg_price`. The DataFrame should only contain one row of values.

```python
#User four previously calculated values into new dataframe.
finalColNames = ['first_reviewed','last_reviewed','nb_private_room','avg_price']
# Variable names oldestReviewDate, newestReviewDate privateRoomListingCount avgListingPrice

rowValues = np.array( [[oldestReviewDate, newestReviewDate, privateRoomListingCount, avgListingPrice]])


finalDF = pd.DataFrame( data = rowValues, columns = finalColNames)
```
